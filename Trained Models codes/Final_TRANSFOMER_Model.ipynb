```python
# ============================================================
# FINAL TRANSFORMER MODEL FOR BEARING FAULT DETECTION
# ============================================================

# !pip -q install tensorflow

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.layers import (
    Input, Dense, Dropout, LayerNormalization, MultiHeadAttention,
    Add, GlobalAveragePooling1D
)
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.models import Model

from google.colab import files
uploaded = files.upload()

# ============================================================
# LOAD DATA
# ============================================================

files_list = os.listdir()

def find_file(keyword):
    matches = [f for f in files_list if keyword in f.lower()]
    return matches[0] if matches else None

train_file = find_file("augmented")
val_file = find_file("validation")
test_file = find_file("test")

if None in [train_file, val_file, test_file]:
    raise FileNotFoundError("Missing required CSV files.")

train_df = pd.read_csv(train_file)
val_df = pd.read_csv(val_file)
test_df = pd.read_csv(test_file)

X_train = train_df.iloc[:, :-1].values
y_train_raw = train_df.iloc[:, -1].values

X_val = val_df.iloc[:, :-1].values
y_val_raw = val_df.iloc[:, -1].values

X_test = test_df.iloc[:, :-1].values
y_test_raw = test_df.iloc[:, -1].values

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)

# Optional original-data augmentation
original_file = None
for f in files_list:
    if "original" in f.lower():
        original_file = f

if original_file is not None:
    original_df = pd.read_csv(original_file)
    X_real = scaler.transform(original_df.iloc[:, :-1].values)
    y_real = original_df.iloc[:, -1].values
    X_train = np.vstack((X_train, X_real))
    y_train_raw = np.concatenate((y_train_raw, y_real))

# Class weights (baseline omitted these; ablation showed +1.54% F1)
classes = np.unique(y_train_raw)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train_raw)
class_weights = dict(zip(classes, weights))

y_train = to_categorical(y_train_raw)
y_val = to_categorical(y_val_raw)
y_test = to_categorical(y_test_raw)

num_features = X_train.shape[1]
X_train = X_train.reshape((X_train.shape[0], num_features, 1))
X_val = X_val.reshape((X_val.shape[0], num_features, 1))
X_test = X_test.reshape((X_test.shape[0], num_features, 1))

# ============================================================
# FINAL TRANSFORMER MODEL
# ============================================================

def transformer_block(x):
    residual = Dense(32)(x)
    x_norm = LayerNormalization()(residual)

    attn = MultiHeadAttention(
        num_heads=4,
        key_dim=16,
        dropout=0.3,
    )(x_norm, x_norm)

    x = Add()([residual, attn])
    x = LayerNormalization()(x)

    ff = Dense(64, activation='relu')(x)
    ff = Dropout(0.3)(ff)
    ff = Dense(32)(ff)

    x = Add()([x, ff])
    return LayerNormalization()(x)

inputs = Input(shape=(num_features, 1))

x = Dense(32, activation="relu")(inputs)

x = transformer_block(x)
x = transformer_block(x)
x = transformer_block(x)

x = GlobalAveragePooling1D()(x)

outputs = Dense(y_train.shape[1], activation='softmax')(x)

model = Model(inputs, outputs)

optimizer = Adam(learning_rate=5e-4)

model.compile(
    optimizer=optimizer,
    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.15),
    metrics=['accuracy']
)

model.summary()

# ============================================================
# CALLBACKS
# ============================================================

callbacks = [
    EarlyStopping(
        monitor='val_accuracy',
        patience=8,
        restore_best_weights=True
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )
]

# ============================================================
# TRAIN FINAL MODEL
# ============================================================

history = model.fit(
    X_train, y_train,
    epochs=70,
    batch_size=64,
    validation_data=(X_val, y_val),
    callbacks=callbacks,
    class_weight=class_weights,
    verbose=1
)

# ============================================================
# EVALUATION
# ============================================================

y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = np.argmax(y_test, axis=1)

accuracy = accuracy_score(y_true, y_pred_classes)
precision = precision_score(y_true, y_pred_classes, average='macro')
recall = recall_score(y_true, y_pred_classes, average='macro')
f1 = f1_score(y_true, y_pred_classes, average='macro')

print("\nFINAL TRANSFORMER PERFORMANCE")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")

# ============================================================
# CONFUSION MATRIX (PLOT + PRINT)
# ============================================================

cm = confusion_matrix(y_true, y_pred_classes)

# Print numerical confusion matrix
print("\nConfusion Matrix (counts):")
print(cm)

# Plot confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens',
            xticklabels=classes, yticklabels=classes)
plt.title("Final Transformer Confusion Matrix")
plt.ylabel("True Label")
plt.xlabel("Predicted Label")
plt.tight_layout()
plt.savefig("final_transformer_confusion_matrix.png", dpi=150)
plt.show()

# Print detailed classification report
print("\nClassification Report:")
print(classification_report(y_true, y_pred_classes, target_names=[str(c) for c in classes]))

# ============================================================
# TRAINING CURVES
# ============================================================

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='val')
plt.title("Final Transformer Accuracy")
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='val')
plt.title("Final Transformer Loss")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("final_transformer_curves.png", dpi=150)
plt.show()

# ============================================================
# SAVE FINAL MODEL
# ============================================================

model.save("final_transformer_model.keras")
print("\nFinal model saved to: final_transformer_model.keras")
```
